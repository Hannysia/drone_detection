```python
import logging
from pathlib import Path

import kagglehub

from src.processing.training import run_full_pipeline

logging.basicConfig(level=logging.INFO, format="%(message)s")

DATASET_ROOT = Path(kagglehub.dataset_download("hannusiaknysh/drone-detection-curated-v1"))
DEFAULT_MODEL_TYPE = "yolo26n.pt"
IMGSZ = 1280
EPOCHS = 50
BATCH_SIZE = 16
NBS = 64
WORKERS = 8
CACHE = "disk"
PATIENCE = 15
BACKUP_EVERY_N_EPOCHS = 10
USE_WANDB = True
REQUIRE_GPU = True
```

## **Execution Block**

### **Positives + Negatives**

```python
run_full_pipeline(
    use_negatives=True,
    run_name="run_with_negatives",
    dataset_root=DATASET_ROOT,
    model_type=DEFAULT_MODEL_TYPE,
    epochs=EPOCHS,
    imgsz=IMGSZ,
    batch=BATCH_SIZE,
    nbs=NBS,
    workers=WORKERS,
    cache=CACHE,
    patience=PATIENCE,
    use_wandb=USE_WANDB,
    backup_every_n_epochs=BACKUP_EVERY_N_EPOCHS,
    require_gpu=REQUIRE_GPU,
)
```

### **Positives-Only**

```python
run_full_pipeline(
    use_negatives=False,
    run_name="run_positives_only",
    dataset_root=DATASET_ROOT,
    model_type=DEFAULT_MODEL_TYPE,
    epochs=EPOCHS,
    imgsz=IMGSZ,
    batch=BATCH_SIZE,
    nbs=NBS,
    workers=WORKERS,
    cache=CACHE,
    patience=PATIENCE,
    use_wandb=USE_WANDB,
    backup_every_n_epochs=BACKUP_EVERY_N_EPOCHS,
    require_gpu=REQUIRE_GPU,
)
```
